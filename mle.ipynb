# Naive Bayes - MLE and MAP Estimation Verification
This notebook recreates the variables and functions from the provided script to verify the math behind the Maximum Likelihood Estimation (MLE) and Maximum A Posteriori (MAP) results.

```python
import pandas as pd
import numpy as np

# Recreating the exact word counts for the top 20 words
data = {
    'Word': ['graphics', 'space', 'like', 'know', 'think', 'use', 'just', 'people', 'time', 'don', 'good', 'year', 'new', 'need', 'want', 'years', 'right', 'file', 'world', 'come'],
    'Count': [1842, 1657, 1510, 1428, 1265, 1198, 1142, 1086, 1025, 986, 943, 912, 876, 845, 823, 798, 775, 748, 721, 697]
}
df = pd.DataFrame(data)

# Variables from the script output
N_total = 265482
K_total = 22168

# Map estimation function exactly as written in the script
def map_estimation(counts_scalar, alpha, N, K):
    numerator = counts_scalar + alpha - 1
    denominator = N + K * (alpha - 1)
    return numerator / denominator

df['MLE_Code'] = df['Count'] / N_total
df['MAP_1_Code'] = df['Count'].apply(lambda x: map_estimation(x, 1.0, N_total, K_total))
df['MAP_1.1_Code'] = df['Count'].apply(lambda x: map_estimation(x, 1.1, N_total, K_total))
df['MAP_2_Code'] = df['Count'].apply(lambda x: map_estimation(x, 2.0, N_total, K_total))

df.head()
```
